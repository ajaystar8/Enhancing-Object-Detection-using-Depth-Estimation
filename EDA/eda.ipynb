## Load data from JSON files

```python
PURE_RGB = os.path.join(".", "final_train_logs", "pure_rgb_final.json")
RGB_HHA_CONCAT_ADDITION_RESNET = os.path.join(".", "final_train_logs", "rgb_hha_resnet_concat_addn_50_0_5_edit.json")
RGB_HHA_CONCAT_DEPTH_RESNET = os.path.join(".", "final_train_logs", "depth_wise_loaded_weight_removed_low_edit.json")
PURE_HHA = None
RGB_HHA_CONCAT_BASIC_CONV = None
```

```python
pure_rgb_data = load_data_from_json_log(PURE_RGB)
pure_hha_data = load_data_from_json_log(PURE_HHA)
rgb_hha_concat_addition_res_data = load_data_from_json_log(RGB_HHA_CONCAT_ADDITION_RESNET)
rgb_hha_concat_depth_res_data = load_data_from_json_log(RGB_HHA_CONCAT_DEPTH_RESNET)
rgb_hha_concat_basic_conv_data = load_data_from_json_log(RGB_HHA_CONCAT_BASIC_CONV)
```

## PURE_RGB

```python
experiment_name = "pure_rgb"
data = pure_rgb_data

plot_loss_epoch_curve(data['loss'], experiment_name)
accuracy_scores = {
    "class_accuracy": data['class_acc'],
    "obj_accuracy": data['obj_acc'],
    "noobj_accuracy": data['noobj_acc']
}
plot_accuracy_epoch_curve(accuracy_scores, experiment_name)
plot_map_epoch_curve(data['map'], "mAP", experiment_name)
plot_ap_per_class(data['ap_per_class'], experiment_name)
```
