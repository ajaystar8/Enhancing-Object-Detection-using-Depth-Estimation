```python
with open("./rgb_1_resume.json", 'r') as f:
    data = json.load(f)

loss = []
for item in data['loss']:
    for key, value in item.items():
        loss.append(value)

data['loss'] = loss

# for item in data.keys():
#     if item == "loss": 
#         continue 
#     for epoch_dict in data[item]: 
#         new_dict = {}
#         for epoch_num, epoch_value in epoch_dict.items(): 
#             epoch_num = int(epoch_num.split('-')[-1]) + 10
#             new_dict[f"Epoch-{epoch_num}"] = epoch_value
#         data[item] = new_dict

for key in data.keys():
    if key == "loss":
        continue
    new_value_list = []
    for epoch_dict in data[key]:
        new_epoch_dict = {}
        for epoch_num, epoch_value in epoch_dict.items():
            epoch_num = int(epoch_num.split('-')[-1]) + 11
            new_epoch_dict[f"Epoch-{epoch_num}"] = epoch_value
            new_value_list.append(new_epoch_dict)
    data[key] = new_value_list

with open('./rgb_1_resume_edit.json', 'w') as f:
    json.dump(data, f)
```

```python
# merge rgb1 and rgb1resume
with open("rgb1.json", 'r') as f:
    rgb_data = json.load(f)

with open("rgb_1_resume_edit.json", 'r') as f:
    rgb_resume_data = json.load(f)

print(rgb_data.keys())
print(rgb_resume_data.keys())

for key in rgb_data.keys():
    if key == "predicted_boxes":
        rgb_data[key] = rgb_resume_data[key]
    rgb_data[key].extend(rgb_resume_data[key])
    

with open("rgb_1_final.json", 'w') as f:
    json.dump(rgb_data, f)
```
